2c. Physiology. In your fixed model, identify the worst stage and explain it from the scoring rules and the signals this dataset actually records (which channels, at what rate, what defines each stage). Show one physiologically expected confusion and one unexpected one.

Ans-
- The lowest classification performance was observed for N1 sleep, with an F1-score of 0.503, compared with 0.979 for wakefulness, 0.845 for N2, 0.887 for N3, and 0.861 for REM.
- This reduced performance is physiologically plausible because N1 represents a transitional state between wakefulness and established sleep, characterised by low-amplitude mixed-frequency EEG activity and slow eye movements. These characteristics can overlap with both wakefulness and N2 sleep, making N1 less distinguishable from neighbouring stages.
- The confusion matrix supports this interpretation, showing misclassifications between N1 and N2 in both directions (49 N1 epochs classified as N2 and 154 N2 epochs classified as N1).
- Another notable confusion involved N1 and REM, with 38 epochs misclassified in each direction.
- Although both stages may exhibit low-amplitude mixed-frequency EEG activity, REM is additionally characterised by rapid eye movements and muscle atonia. The model uses two EEG derivations (Fpz-Cz and Pz-Oz) and horizontal EOG, sampled at 100 Hz, but does not use the recorded submental EMG channel.
- Consequently, it cannot directly exploit chin muscle activity to distinguish REM from other sleep stages.
- Overall, the results indicate that classification difficulty is associated not only with model performance but also with the physiological similarity and transitional characteristics of individual sleep stages.
